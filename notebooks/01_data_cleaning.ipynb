# Data Cleaning


```python
# import necessary dependancies
import pandas as pd
import matplotlib.pyplot as plt
from glob import glob
```

## Get data

```python
def get_data(data):
    datasets =glob(data)
    files=[]
    for dataset in datasets:
        data=pd.read_csv(dataset)
        files.append(data)
    
    return pd.concat(files,ignore_index=True)  

df=get_data("../data/buenos-aires-real-estate-*.csv")
```

## Clean data

```python
cleaned_df=(
    df.assign(
        state=lambda x : x["place_with_parent_names"].str.split("|").str[2].astype(str),
         lat=lambda x:x["lat-lon"].str.split(",").str[0].astype(float),
         lon=lambda x:x["lat-lon"].str.split(",").str[1].astype(float)
    ).drop(columns=["place_with_parent_names","lat-lon"])
    .rename(columns={"properati_url":"property_url"})
    
       
    
)
```
